```python
# Nhập mảng bằng cách nhập các số cách nhau bởi dấu cách
try:
    a = list(map(int, input("Nhập các phần tử mảng a (cách nhau bởi dấu cách): ").split()))
    b = list(map(int, input("Nhập các phần tử mảng b (cách nhau bởi dấu cách): ").split()))

    def tron_mang_min(a, b):
        """
        Trộn 2 mảng a và b thành mảng mới:
        - Mỗi phần tử là min(a[i], b[i])
        - Nếu còn phần tử dư ở a hoặc b, thêm vào mảng mới
        """
        c = []
        n = min(len(a), len(b))

        for i in range(n):
            c.append(min(a[i], b[i]))

        c.extend(a[n:])
        c.extend(b[n:])
        return c

    result = tron_mang_min(a, b)
    print("Mảng sau khi trộn:", result)
except ValueError:
    print("Lỗi: Vui lòng chỉ nhập các số nguyên cách nhau bởi dấu cách.")
```

```python
import random

def tao_ma_tran(n, min_val=-10, max_val=10):
    """Tạo ma trận vuông n x n với số nguyên ngẫu nhiên"""
    return [[random.randint(min_val, max_val) for j in range(n)] for i in range(n)]

def xuat_ma_tran(a):
    for row in a:
        print(row)

# Ví dụ
n = 4
a = tao_ma_tran(n)
print("Ma trận a:")
xuat_ma_tran(a)

def tong_tam_giac_tren_phu(a):
    n = len(a)
    total = 0
    for i in range(n):
        for j in range(n):
            if j >= n-1-i:  # nằm trên hoặc trên đường chéo phụ
                total += a[i][j]
    return total
def chuyen_am_thanh_abs(a):
    n = len(a)
    for i in range(n):
        for j in range(n):
            if a[i][j] < 0:
                a[i][j] = abs(a[i][j])
def toan_chan(a):
    n = len(a)
    for i in range(n):
        for j in range(n):
            if a[i][j] % 2 != 0:
                return False
    return True
def doi_xung(a):
    n = len(a)
    for i in range(n):
        for j in range(i+1, n):  # chỉ cần kiểm tra nửa trên
            if a[i][j] != a[j][i]:
                return False
    return True
def duong_cheo_chinh_tang(a):
    n = len(a)
    for i in range(n-1):
        if a[i][i] >= a[i+1][i+1]:
            return False
    return True
def tam_giac_duoi_phu(a):
    n = len(a)
    result = []
    for i in range(n):
        for j in range(n):
            if j <= n-1-i:  # nằm dưới hoặc trên đường chéo phụ
                result.append(a[i][j])
    return result
def duong_cheo_phu_giam(a):
    n = len(a)
    for i in range(n-1):
        if a[i][n-1-i] <= a[i+1][n-2-i]:
            return False
    return True
```
