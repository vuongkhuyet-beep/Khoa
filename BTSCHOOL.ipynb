```python
def tron_mang(a, b):
    c = []
    n = min(len(a), len(b))  # số phần tử chung nhỏ nhất

    # Trộn phần tử tương ứng
    for i in range(n):
        c.append(a[i] + b[i])

    # Thêm phần tử còn lại của a (nếu có)
    for i in range(n, len(a)):
        c.append(a[i])

    # Thêm phần tử còn lại của b (nếu có)
    for i in range(n, len(b)):
        c.append(b[i])

    return c
```

```python
import random

def tao_mang(m, n, min_val=0, max_val=100):
    """Tạo mảng m x n chứa số nguyên ngẫu nhiên từ min_val đến max_val"""
    a = [[random.randint(min_val, max_val) for j in range(n)] for i in range(m)]
    return a

def xuat_dong(a, k):
    """Trả về dòng k của ma trận a"""
    if 0 <= k < len(a):
        return a[k]
    else:
        return "Chỉ số dòng không hợp lệ"
def xuat_cot(a, k):
    """Trả về cột k của ma trận a"""
    if 0 <= k < len(a[0]):
        return [row[k] for row in a]
    else:
        return "Chỉ số cột không hợp lệ"

def dong_tong_lon_nhat(a):
    max_sum = sum(a[0])
    max_row_index = 0
    for i, row in enumerate(a):
        if sum(row) > max_sum:
            max_sum = sum(row)
            max_row_index = i
    return max_row_index, max_sum

def cot_tich_nho_nhat(a):
    n = len(a[0])
    min_product = None
    min_col_index = -1
    for j in range(n):
        product = 1
        for i in range(len(a)):
            product *= a[i][j]
        if min_product is None or product < min_product:
            min_product = product
            min_col_index = j
    return min_col_index, min_product

def dong_chan_cot_le(a):
    result = []
    for i in range(len(a)):
        if i % 2 == 0:  # dòng chẵn (0,2,4...)
            for j in range(len(a[0])):
                if j % 2 == 1:  # cột lẻ (1,3,5...)
                    result.append(a[i][j])
    return result

# Example usage:
a = tao_mang(3, 4, 1, 10)
print("Ma trận A:", a)
print("Phần tử dòng chẵn, cột lẻ:", dong_chan_cot_le(a))

def trung_binh_dong_le_chi_so_chan(a):
    total = 0
    count = 0
    for i in range(len(a)):
        if i % 2 == 1:  # dòng lẻ (1,3,...)
            for val in a[i]:
                if val % 2 == 0:
                    total += val
                    count += 1
    return total / count if count > 0 else 0

print(trung_binh_dong_le_chi_so_chan(a))

def trung_binh_bien(a):
    total = 0
    count = 0
    m = len(a)
    n = len(a[0])
    for i in range(m):
        for j in range(n):
            if i == 0 or i == m-1 or j == 0 or j == n-1:
                total += a[i][j]
                count += 1
    return total / count if count > 0 else 0

print(trung_binh_bien(a))

def trung_binh_khong_bien(a):
    total = 0
    count = 0
    m = len(a)
    n = len(a[0])
    for i in range(1, m-1):
        for j in range(1, n-1):
            total += a[i][j]
            count += 1
    return total / count if count > 0 else 0

print(trung_binh_khong_bien(a))
```

```python
class SinhVien:
    def __init__(self, ma, ten, nam_sinh, diem_tb):
        self.ma = ma              # chuỗi 10 ký tự
        self.ten = ten            # chuỗi tối đa 20 ký tự
        self.nam_sinh = nam_sinh  # số nguyên
        self.diem_tb = diem_tb    # số thực

    def __str__(self):
        return f"{self.ma} | {self.ten} | {self.nam_sinh} | {self.diem_tb}"
n = int(input("Nhập số sinh viên: "))

ds_sv = []

for i in range(n):
    print(f"Nhập thông tin sinh viên thứ {i+1}:")
    ma = input("Mã sinh viên (10 ký tự): ")
    ten = input("Tên (tối đa 20 ký tự): ")
    nam_sinh = int(input("Năm sinh: "))
    diem_tb = float(input("Điểm trung bình: "))
    sv = SinhVien(ma, ten, nam_sinh, diem_tb)
    ds_sv.append(sv)
def dem_sv_len_lop(ds_sv):
    return sum(1 for sv in ds_sv if sv.diem_tb >= 5)

print("Số sinh viên đủ điều kiện lên lớp:", dem_sv_len_lop(ds_sv))
def xuat_sv_20_tuoi(ds_sv, nam_hien_tai=2026):
    for sv in ds_sv:
        if nam_hien_tai - sv.nam_sinh == 20:
            print(sv)

print("Sinh viên đủ 20 tuổi:")
xuat_sv_20_tuoi(ds_sv)
def dem_sv_DH(ds_sv):
    return sum(1 for sv in ds_sv if sv.ma[2:4] == "DH")

print("Số sinh viên hệ đại học:", dem_sv_DH(ds_sv))
def dem_sv_DH(ds_sv):
    return sum(1 for sv in ds_sv if sv.ma[2:4] == "DH")

print("Số sinh viên hệ đại học:", dem_sv_DH(ds_sv))
def dem_sv_ten_lan(ds_sv):
    return sum(1 for sv in ds_sv if sv.ten.split()[-1] == "Lan")

print("Số sinh viên tên Lan:", dem_sv_ten_lan(ds_sv))
def dem_sv_ho_phan(ds_sv):
    return sum(1 for sv in ds_sv if sv.ten.split()[0] == "Phan")

print("Số sinh viên họ Phan:", dem_sv_ho_phan(ds_sv))
```
