```python
import heapq

N = 4

def manhattan(state, goal):
    dist = 0
    for i in range(N*N):
        if state[i] == 0:
            continue
        x1, y1 = divmod(i, N)
        x2, y2 = divmod(goal.index(state[i]), N)
        dist += abs(x1 - x2) + abs(y1 - y2)
    return dist

def neighbors(state):
    res = []
    idx = state.index(0)
    x, y = divmod(idx, N)

    moves = [(-1,0),(1,0),(0,-1),(0,1)]

    for dx, dy in moves:
        nx, ny = x+dx, y+dy
        if 0 <= nx < N and 0 <= ny < N:
            new_idx = nx*N + ny
            new_state = list(state)
            new_state[idx], new_state[new_idx] = new_state[new_idx], new_state[idx]
            res.append(tuple(new_state))
    return res

def a_star_15puzzle(start, goal):
    open_set = []
    heapq.heappush(open_set, (0, start))

    g = {start: 0}
    parent = {start: None}

    while open_set:
        _, current = heapq.heappop(open_set)

        if current == goal:
            path = []
            while current:
                path.append(current)
                current = parent[current]
            return path[::-1]

        for nxt in neighbors(current):
            temp_g = g[current] + 1

            if nxt not in g or temp_g < g[nxt]:
                g[nxt] = temp_g
                f = temp_g + manhattan(nxt, goal)
                heapq.heappush(open_set, (f, nxt))
                parent[nxt] = current

    return None
```

```python
import heapq

def a_star_graph(graph, start, goal, h):
    open_set = []
    heapq.heappush(open_set, (0, start))

    g = {start: 0}
    parent = {start: None}

    while open_set:
        _, current = heapq.heappop(open_set)

        if current == goal:
            path = []
            while current:
                path.append(current)
                current = parent[current]
            return path[::-1], g[goal]

        for neighbor, cost in graph[current]:
            temp_g = g[current] + cost

            if neighbor not in g or temp_g < g[neighbor]:
                g[neighbor] = temp_g
                f = temp_g + h(neighbor)
                heapq.heappush(open_set, (f, neighbor))
                parent[neighbor] = current

    return None, float('inf')
```
